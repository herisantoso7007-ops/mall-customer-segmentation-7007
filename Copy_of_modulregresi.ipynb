```python
import pandas as pd

url = 'https://raw.githubusercontent.com/qtle3/simple-linear-regression/refs/heads/main/Salary_Data.csv'
df = pd.read_csv(url)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='YearsExperience', y='Salary')
plt.title('Hubungan antara Tahun Pengalaman dan Gaji')
plt.xlabel('Tahun Pengalaman')
plt.ylabel('Gaji')
plt.grid(True)
plt.show()

print("Visualisasi di atas menunjukkan scatter plot yang menggambarkan hubungan antara tahun pengalaman kerja dan gaji. Dapat terlihat jelas adanya tren positif yang kuat, di mana seiring bertambahnya tahun pengalaman, gaji cenderung meningkat. Titik-titik data sebagian besar mengikuti pola linear, menunjukkan bahwa model regresi linear mungkin cocok untuk data ini.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
import numpy as np

# 1. Menyiapkan data
X = df['YearsExperience'].values.reshape(-1, 1)  # Fitur (independen)
y = df['Salary'].values.reshape(-1, 1)         # Target (dependen)

# 2. Membagi data menjadi data latih dan data uji
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# 3. Membuat dan melatih model Regresi Linear Sederhana
regressor = LinearRegression()
regressor.fit(X_train, y_train) #membuat model regresi

# 4. Membuat prediksi pada data uji
y_pred = regressor.predict(X_test)

# 5. Menghitung metrik error
mae = metrics.mean_absolute_error(y_test, y_pred)
mse = metrics.mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = metrics.r2_score(y_test, y_pred)

# 6. Mendapatkan persamaan regresi
intercept = regressor.intercept_[0]
slope = regressor.coef_[0][0]

print("\n--- Hasil Analisis Regresi Linear Sederhana ---")
print(f"Intercept (konstanta): {intercept:.2f}")
print(f"Koefisien (slope) YearsExperience: {slope:.2f}")

print(f"Persamaan Regresi: Gaji = {slope:.2f} * YearsExperience + {intercept:.2f}")

print("\n--- Metrik Evaluasi Model ---")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared (R2): {r2:.2f}")

print("\n--- Penjelasan Hasil ---")
print("**Intercept (Konstanta)**: Nilai ini merupakan perkiraan gaji ketika YearsExperience (tahun pengalaman) adalah 0. Dalam kasus ini, nilai Intercept sekitar Rp {intercept:.2f}.")
print("**Koefisien (Slope)**: Nilai ini menunjukkan seberapa besar perubahan gaji untuk setiap penambahan satu tahun pengalaman. Koefisien YearsExperience sebesar Rp {slope:.2f} berarti bahwa untuk setiap tambahan satu tahun pengalaman kerja, gaji diperkirakan akan meningkat sekitar Rp {slope:.2f}.")
print("**Persamaan Regresi**: Model yang ditemukan adalah Gaji = {slope:.2f} * YearsExperience + {intercept:.2f}. Persamaan ini dapat digunakan untuk memprediksi gaji berdasarkan tahun pengalaman.")
print("\n**Metrik Evaluasi**:")
print("  - **Mean Absolute Error (MAE)**: Rata-rata selisih absolut antara nilai prediksi dan nilai aktual. MAE sebesar {mae:.2f} menunjukkan bahwa rata-rata kesalahan prediksi gaji adalah sekitar Rp {mae:.2f}.")
print("  - **Mean Squared Error (MSE)**: Rata-rata kuadrat selisih antara nilai prediksi dan nilai aktual. MSE lebih sensitif terhadap outlier. Nilai {mse:.2f} adalah rata-rata kuadrat error.")
print("  - **Root Mean Squared Error (RMSE)**: Akar kuadrat dari MSE. RMSE sebesar {rmse:.2f} memberikan indikasi ukuran kesalahan dalam unit yang sama dengan variabel target (gaji).")
print("  - **R-squared (R2)**: Proporsi varians dalam variabel dependen yang dapat diprediksi dari variabel independen. Nilai R2 sebesar {r2:.2f} (atau {r2*100:.2f}%) menunjukkan bahwa sekitar {r2*100:.2f}% dari variasi gaji dapat dijelaskan oleh tahun pengalaman. Nilai R2 yang mendekati 1 menunjukkan model yang sangat baik dalam menjelaskan variasi data.")
print("Secara keseluruhan, model regresi linear sederhana ini terlihat cukup baik dalam menjelaskan hubungan antara tahun pengalaman dan gaji, dengan tahun pengalaman memiliki pengaruh positif yang signifikan terhadap gaji.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
import numpy as np

# 1. Menyiapkan data untuk Multiple Linear Regression (df2)
# Fitur (variabel independen): TV, radio, newspaper
X_mlr = df2[['TV', 'radio', 'newspaper']]
# Target (variabel dependen): sales
y_mlr = df2['sales']

# 2. Membagi data menjadi data latih dan data uji
# Menggunakan 80% data untuk pelatihan dan 20% untuk pengujian
X_train_mlr, X_test_mlr, y_train_mlr, y_test_mlr = train_test_split(X_mlr, y_mlr, test_size=0.2, random_state=42)

print("Ukuran data training (X_train_mlr):", X_train_mlr.shape)
print("Ukuran data testing (X_test_mlr):", X_test_mlr.shape)

# 3. Membuat dan melatih model Regresi Linear Berganda
regressor_mlr = LinearRegression()
regressor_mlr.fit(X_train_mlr, y_train_mlr)

# 4. Membuat prediksi pada data uji
y_pred_mlr = regressor_mlr.predict(X_test_mlr)

# 5. Menghitung metrik error
mae_mlr = metrics.mean_absolute_error(y_test_mlr, y_pred_mlr)
mse_mlr = metrics.mean_squared_error(y_test_mlr, y_pred_mlr)
rmse_mlr = np.sqrt(mse_mlr)
r2_mlr = metrics.r2_score(y_test_mlr, y_pred_mlr)

# 6. Mendapatkan persamaan regresi
intercept_mlr = regressor_mlr.intercept_
coefficients_mlr = regressor_mlr.coef_

print("\n--- Hasil Analisis Regresi Linear Berganda ---")
print(f"Intercept (konstanta): {intercept_mlr:.2f}")
print("Koefisien untuk setiap fitur:")
for feature, coef in zip(X_mlr.columns, coefficients_mlr):
    print(f"  {feature}: {coef:.2f}")

print(f"Persamaan Regresi: Sales = {intercept_mlr:.2f} + {coefficients_mlr[0]:.2f}*TV + {coefficients_mlr[1]:.2f}*radio + {coefficients_mlr[2]:.2f}*newspaper")

print("\n--- Metrik Evaluasi Model ---")
print(f"Mean Absolute Error (MAE): {mae_mlr:.2f}")
print(f"Mean Squared Error (MSE): {mse_mlr:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse_mlr:.2f}")
print(f"R-squared (R2): {r2_mlr:.2f}")

print("\n--- Penjelasan Hasil Analisis Regresi Linear Berganda ---")
print("Model ini mencoba memprediksi 'Sales' berdasarkan pengeluaran iklan di 'TV', 'radio', dan 'newspaper'.")
print("\n**Intercept (Konstanta)**: Nilai ini mewakili perkiraan penjualan ketika semua pengeluaran iklan (TV, radio, koran) adalah nol. Dalam kasus ini, Intercept sekitar {intercept_mlr:.2f}.")
print("\n**Koefisien Regresi**:")
print(f"  - **TV**: Koefisien {coefficients_mlr[0]:.2f} menunjukkan bahwa untuk setiap peningkatan 1 unit pengeluaran iklan TV (misal: 1000 dolar), penjualan diperkirakan akan meningkat sebesar {coefficients_mlr[0]:.2f} unit (dengan asumsi pengeluaran radio dan koran konstan). Koefisien positif yang tinggi menunjukkan dampak yang signifikan.")
print(f"  - **Radio**: Koefisien {coefficients_mlr[1]:.2f} menunjukkan bahwa untuk setiap peningkatan 1 unit pengeluaran iklan radio, penjualan diperkirakan akan meningkat sebesar {coefficients_mlr[1]:.2f} unit.")
print(f"  - **Newspaper**: Koefisien {coefficients_mlr[2]:.2f} menunjukkan bahwa untuk setiap peningkatan 1 unit pengeluaran iklan koran, penjualan diperkirakan akan berubah sebesar {coefficients_mlr[2]:.2f} unit. Perhatikan apakah nilai ini positif atau negatif, dan seberapa besar dampaknya dibandingkan dengan TV dan radio.")
print("\n**Persamaan Regresi**: Persamaan Sales = {intercept_mlr:.2f} + {coefficients_mlr[0]:.2f}*TV + {coefficients_mlr[1]:.2f}*radio + {coefficients_mlr[2]:.2f}*newspaper dapat digunakan untuk memprediksi nilai penjualan berdasarkan nilai pengeluaran iklan.")
print("\n**Metrik Evaluasi**:")
print(f"  - **Mean Absolute Error (MAE)**: Rata-rata selisih absolut antara prediksi dan nilai sebenarnya adalah {mae_mlr:.2f}. Ini mengukur rata-rata kesalahan dalam unit penjualan.")
print(f"  - **Mean Squared Error (MSE)**: Rata-rata kuadrat kesalahan adalah {mse_mlr:.2f}. Memberikan penekanan lebih pada kesalahan yang lebih besar.")
print(f"  - **Root Mean Squared Error (RMSE)**: RMSE sebesar {rmse_mlr:.2f} memberikan ukuran kesalahan dalam unit yang sama dengan 'Sales', sehingga lebih mudah diinterpretasikan.")
print(f"  - **R-squared (R2)**: Nilai R2 sebesar {r2_mlr:.2f} (atau {r2_mlr*100:.2f}%) menunjukkan bahwa sekitar {r2_mlr*100:.2f}% dari variasi dalam penjualan dapat dijelaskan oleh model ini menggunakan pengeluaran iklan TV, radio, dan koran. Nilai yang mendekati 1 menunjukkan model yang sangat baik dalam menjelaskan variasi data.")
print("Secara keseluruhan, model regresi linear berganda ini menunjukkan bagaimana berbagai saluran iklan berkontribusi terhadap penjualan, dengan TV dan Radio umumnya memiliki dampak positif yang lebih signifikan dibandingkan Newspaper.")
```

```python
import gradio as gr
import numpy as np
import pandas as pd # Import pandas for DataFrame creation if needed, though numpy array also works

# Pastikan regressor_mlr, intercept_mlr, dan coefficients_mlr sudah terdefinisi dari sel sebelumnya
# Jika tidak, Anda perlu menjalankan sel sebelumnya atau memuat ulang model jika ini adalah sesi baru.
# Asumsi variabel-variabel ini sudah ada di lingkungan kernel.

# Fungsi prediksi yang akan digunakan oleh Gradio
def predict_sales(tv, radio, newspaper):
    # Buat array numpy dari input pengguna
    features = np.array([[tv, radio, newspaper]])

    # Lakukan prediksi menggunakan model yang sudah dilatih
    predicted_sales = regressor_mlr.predict(features)[0]

    return predicted_sales

# Buat antarmuka Gradio
iface = gr.Interface(
    fn=predict_sales,
    inputs=[
        gr.Number(label="Pengeluaran TV (dalam ribuan dolar)", value=150),
        gr.Number(label="Pengeluaran Radio (dalam ribuan dolar)", value=20),
        gr.Number(label="Pengeluaran Koran (dalam ribuan dolar)", value=10)
    ],
    outputs=gr.Number(label="Prediksi Penjualan (dalam unit ribuan)"),
    title="Prediksi Penjualan Berdasarkan Pengeluaran Iklan",
    description="Masukkan jumlah pengeluaran di TV, radio, dan koran untuk memprediksi penjualan."
)

# Jalankan antarmuka Gradio
iface.launch()
```
